```python
class ExpertPlacement:
  def __init__(self,num_experts:int,num_gpus:int):
    if num_experts <= 0 :
      raise ValueError("Number of experts need to be greater than 0")
    if num_gpus <= 0 :
      raise ValueError("Number of GPU needs to be ")
    if num_experts%num_gpus != 0:
      raise ValueError("It requires equal number of experts for each GPU")

    self.num_experts = num_experts
    self.num_gpus = num_gpus
    self.num_experts_per_gpu = num_experts//num_gpus

    #map every expert to it's own GPU
    self.experts_to_gpu = {}

    #map every GPU to it's local expert
    self.gpu_to_experts = {}

    for expert_id in range(num_experts):
      gpu_id = expert_id // self.num_experts_per_gpu
      self.experts_to_gpu[expert_id] = gpu_id
      self.gpu_to_experts.setdefault(gpu_id,[]).append(expert_id)


  def get_owner(self,expert_id: int)-> int:
    if expert_id not in self.experts_to_gpu:
      raise ValueError("Invalid expert id")
    return self.experts_to_gpu[expert_id]

  def get_experts(self,gpu_id:int)->list:
    if gpu_id  not in self.gpu_to_experts:
      raise ValueError("Invalid GPU id")
    return self.gpu_to_experts[gpu_id]

  def print_placement(self):

        print(f"Number of experts: {self.num_experts}")

        print(f"Number of GPUs: {self.num_gpus}")

        print(f"Experts per GPU: {self.experts_per_gpu}")

        print("\nExpert Placement:")

        for gpu_id, experts in self.gpu_to_experts.items():

            print(f"GPU {gpu_id}: {experts}")



placement = ExpertPlacement(num_experts=8,num_gpus=4)

placement.print_placement()

print("\nOwner of Expert 5:", placement.get_owner(5))

print("Experts on GPU 2:", placement.get_local_experts(2))
```
